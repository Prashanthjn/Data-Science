# mode using statistics library

```python
import seaborn as sns
import statistics
```

```python
data=sns.load_dataset('tips')
statistics.mode(data['total_bill'])
```

# variance 

```python
sample_var=np.var(data['total_bill'], ddof=1)
pop_var=np.var(data['total_bill'], ddof=0)
print(sample_var)
print(pop_var)
```

```python
series=pd.Series(data['total_bill'])
#print(series)
samp_var=series.var()
print('sample variace :',samp_var)
pop_var=series.var(ddof=0)
print('population variance:',pop_var)
```

# using scipy
