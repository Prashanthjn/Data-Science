2. Measure of variability

```python
import numpy as np

arr=[2,3,4,6,6]

mini=min(arr)
maxi=max(arr)

ran=maxi-mini

print(ran)
```

```python
import statistics
arr=[2,3,4,6,6]
var=statistics.variance(arr)
print(var)
```

```python
import statistics
arr=[2,3,4,6,6]
std_dev=statistics.stdev(arr)
print(std_dev)
```
