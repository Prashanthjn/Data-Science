# handling missing data

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]      #nan means a null value
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
df.isna()     #to check if the value is null
```

```python
df.isna().sum()   #to find the sum of null values
```

```python
df.isna().any()      #tp find if any null values exist in each row
```

```python
df.dropna()    #to drop the rows with null values
```

```python
df.dropna(thresh=3)
```
