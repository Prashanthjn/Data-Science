# Dataframes
* it's just the collection of multiple of series

```python
import numpy as np
import pandas as pd
```

Dataframe from dictionary

```python
data={
    'name':['ace','bob','carry','danny','emma'],
    'age':[20,30,20,30,20],
    'usn':['cd041','cd031','cd023','cd043','cd082']
}
```

```python
df=pd.DataFrame(data)
df
```

dataframe from list

```python
li=[
    ['john','cd032'],
    ['jack','cd031'],
    ['jerry','cd035']
    ]
cols=['name','usn']
df1=pd.DataFrame(li)          #here the headers will be from 0 to n
print(df1)
df2=pd.DataFrame(li,columns=cols)   # setting labels
df2
```
