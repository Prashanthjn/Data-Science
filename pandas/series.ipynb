# series
* can store any type of data it's stacked in vertical order

```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c']
my_list=[1,2,3]
ar=np.array([4,5,6])
d={1:10,2:20,3:30}
```

```python
list_Series=pd.Series(my_list,index=labels)  #labels is optional and default labels are 0 to n
list_Series
```

```python
array_series=pd.Series(ar,index=labels)
array_series
```

```python
dict_series=pd.Series(d)  #here we should not give any indexxes since the dict will have default keys 
dict_series
```
