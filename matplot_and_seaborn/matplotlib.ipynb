```python
import numpy as np
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,20,10)
y=x**2

print(x)
print(y)
```

# basic matplotlib

```python
plt.plot(x,y)
plt.title('square graph')
plt.xlabel('x=x')
plt.ylabel('y=x^2')
```

# subplots in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,x*3)
plt.subplot(2,2,2)
plt.plot(x,x**2)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y**3)
```

```python
fig=plt.figure(figsize=(10,5),dpi=250)
axis1=fig.add_axes([0.1,0.1,.4,.8])
axis1.plot(x,y)
axis2=fig.add_axes([.1,.4,.4,.6])
axis2.plot(x,x**3)
```
