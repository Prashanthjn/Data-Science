```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
flight=sns.load_dataset('flights')
flight.head()
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
tipcorr=tips[['total_bill','tip','size']]
tipcorr
```

```python
sns.heatmap(tipcorr.corr(),annot=True)
```

```python
sns.clustermap(tipcorr.corr(),annot=True)
```

```python
import numpy as np
import pandas as pd
```

```python
pivotflight=flight.pivot_table(values='passengers',index='month',columns='year')
```

```python
pivotflight
```

```python
sns.heatmap(pivotflight)
```

# regression plot

```python
tips = sns.load_dataset('tips')
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips)
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex')
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',palette='rainbow')
```

```python
sns.lmplot(x='total_bill',y='tip',data=tips,hue='sex',palette='rainbow',markers=['o','p'],scatter_kws={'s':50})
```
