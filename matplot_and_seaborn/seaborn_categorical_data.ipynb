# these are used to plot the categorical data
* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=sns.load_dataset('tips')
```

```python
df.head()
```

```python
sns.countplot(x=df['smoker'],hue=df['sex'])
```

```python
sns.barplot(x=df['sex'],y=df['tip'],estimator=np.median) #shows the mean as default estimator
```

```python
sns.boxplot(x=df['tip'],y=df['day'])   #x takes the qualitative value and y takes quantitative 
```

```python
sns.violinplot(x=df['day'],y=df['tip'])
```

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```
