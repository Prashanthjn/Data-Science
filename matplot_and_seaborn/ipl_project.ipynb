```python
match_wins=data['match_winner'].value_counts()
sns.barplot(y=match_wins.index,x=match_wins.values,palette='rainbow')
plt.title('match win analysis')
plt.xlabel('number of matches won')
plt.ylabel('teams')
```

```python
toss=data['toss_decision']
toss.value_counts()
sns.countplot(x=toss,palette='rainbow')
```

```python
count=data[data['toss_winner']==data['match_winner']]['match_id'].count()
count
percentage=(count*100)/data.shape[0]
print(f'probability that toss winner will also win match is {percentage.round(2)}')
```

```python
potm=data['player_of_the_match'].value_counts().head(10)
sns.barplot(x=potm.values,y=potm.index,palette='mako')
plt.title('most player of the match awards')
```

```python
topscore=data.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(5)
topscore.plot(kind='bar')
```

```python
data['highest_wickets']=data['best_bowling_figure'].apply(lambda x:x.split('--')[0]).astype(int)
bestbowl=data.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(5)
bestbowl.plot(kind='bar')
```
