# broadcasting

```python

#in broadcasting the values of rows or columns are being broadcasted 
```

```python
ar1=np.arange(1,21).reshape(5,4)
ar1
```

```python
ar2=ar1.copy()
ar2
ar2=ar2*3              #multiplies all the values by 3
print(ar1)             #so here the 3 is broadcastedto all the elements
print(ar2)
```

```python
a1=np.array([2,3,4])
b1=np.array([
    [1],
    [2],
    [3]
    ])
print(a1)
print(b1)
```

```python
print(a1*b1)  
#here also the values will be automatically broadcasted for multiplication
```

# deep copy and shallow copy
* shallow copy creates a new object but 
changes made in nested element "affects" the original
* deep copy creates a independent copy of the object and all the 
changes made in nested objec "doesn't affect" the original original

```python
arr1=np.arange(1,21)
arr1
```

```python
# shallow copy  affects the original array
arr3=arr1
arr3[5]=100
print(arr3)
print(arr1)
```

```python
#deep copy  doesn't affect the original
arr2=arr1[:5]
arr2=arr2*3
print(arr2)
print(arr1)
print(arr3)   #arr1 and arr3 both refer to same address
```
