# vector(1D array) indexing and slicing

```python
arr=np.arange(11,21)
arr
```

```python
arr[5]
```

```python
arr[1:3]
```

```python
arr[::-1]
```

# numpy indexing and slicing of matrix

```python
ar1=np.arange(1,31).reshape(5,6)
ar1
```

```python
ar1[0]
```

```python
ar1[2,4]
```

```python
ar1[0:2,3:5]
```

```python
ar1[2:5,4:6]
```

```python
ar1[:,3]    #to selct full single column
```

```python
ar1[3,:]   # to select full single row 
```

# boolean indexing

```python
arr=np.arange(20).reshape(4,5)
result=arr[(arr>5)& (arr<10)]
print(arr)
print(result)
res=arr[arr%2==0]
print(res)
```
