```python
from collections import deque

deq=deque([1,2,3])

deq.append(5)         #deque is double ended que you can insert from both sides  
deq.appendleft(6)

print(deq)
deq.reverse()
print(deq)
```

```python
deq.pop()
print(deq)
deq.popleft()
print(deq)
```
