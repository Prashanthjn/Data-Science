```python
from collections import UserString

class strr(UserString):
    def __init__(self, seq):
        self.data=seq.upper()
    
    def append(self,value):
        self.data+=value

    def remove(self,rem):
        self.data=self.data.replace(rem,"")
s= strr('prashi ')
print(s)
s.append("j n ")
print(s)
s.remove("P")
print(s)
```
