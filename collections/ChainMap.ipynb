```python
from collections import ChainMap
d1={'a':1,'b':3}
d2={'c':3,'d':4}

c=ChainMap(d1,d2)

for key,value in c.items():
    print(key,value)
    
print(c)

print(c.values())
print(c.keys())
```

```python
from collections import ChainMap

d1={'a':1,'b':3}
d2={'c':3,'d':4}
d3={'e':5}

c=ChainMap(d1,d2)

f=c.new_child(d3)

print(c)
print(f)
```
