```python
from collections import defaultdict

d=defaultdict(float)

l=[1,2,3,4,5]

for i in range(2):
    d['a']+=3

print(d)
print(l)
```

```python
from collections import defaultdict

cnt=defaultdict(int)

fruits=['apple','mango','mango']

for fruit in fruits:
    cnt[fruit]+=1
print(cnt)
```
