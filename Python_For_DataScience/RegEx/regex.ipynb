```python
import re

s="""the number is 24245353245 
    another number is 3859385"""
regex='\d+'
REX='\d'
mat=re.findall(REX,s)
print(mat)
match=re.findall(regex,s)
print(match)
```

```python
import re
s="""the number is 24245353245 
    another number is 3859385 $% * """

p=re.compile('\w')
p1=re.compile('\w+')
p2=re.compile('\W')

print(p.findall(s))
print(p1.findall(s))
print(p2.findall(s))
```

```python
from re import split

print(split('\W+', 'On 12th Jan 2016, at 11:02 AM'))
print(split('\d+', 'On 12th Jan 2016, at 11:02 AM'))
```
