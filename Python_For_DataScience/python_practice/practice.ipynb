```python
def large(nums):
    lar=nums[0]
    for i in range(1,len(nums)):
        if nums[i]>lar:
            lar=nums[i]
    return lar
large([5,9,2,8,1])
```

```python
def small(nums):
    smal=nums[0]
    for i in range(1,len(nums)):
        if nums[i]<smal:
            smal=nums[i]
    return smal
small([5,9,2,8,1])
```

```python
def count_even(nums):
    count=0
    for i in nums:
        if i%2==0:
            count+=1
    return count
count_even([5,9,2,8,1])
```

```python
def sum_list(nums):
    sum=0
    for i in nums:
        sum=sum+i
    return sum
sum_list([5,9,2,8,1])
```

```python
def avg_list(nums):
    avg,sum=0,0
    for i in nums:
        sum=sum+i
    avg=sum/len(nums)
    return avg
avg_list([5,9,2,8,1])
```
