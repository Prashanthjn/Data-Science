```python
import numpy as np 
from scipy import stats 

b = np.array([120, 122, 118, 130, 125, 128, 115, 121, 123, 119])
a = np.array([115, 120, 112, 128, 122, 125, 110, 117, 119, 114])

alpha=0.05

t_stat,p_val=stats.ttest_rel(a,b)

m=np.mean(a-b)
s=np.std(a-b,ddof=1)
n=len(b)
t_manual=m/(s/np.sqrt(n))

decision= "reject" if p_val<= alpha else "fail to reject"
concl=" significance diffrenece" if decision == "reject" else "no significance diffrence"

print("T:", t_stat)
print("P:",p_val)
print("T manual:",t_manual)
print(f"Decision: {decision} H0 at alpha={alpha}")
print(f"Conclusion:{concl}")


#Steps for the hypothesis testing

    #          Hypothesis Testing
    #                  ↓
    #          Define H₀ and H₁
    #                  ↓
    #          Choose α = 0.05
    #                  ↓
    #           Collect Sample
    #                  ↓
    #          Choose Statistical Test
    #                  ↓
    #       Calculate Test Statistic
    #                  ↓
    #              Find p-value
    #                  ↓
    #           Compare p with α
    #             ↙           ↘
    #       p ≤ α              p > α
    #          ↓                  ↓
    #    Reject H₀          Fail to Reject H₀      
```
