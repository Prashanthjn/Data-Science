```python
import numpy as np
import seaborn as sns
```

```python
arr=np.array([1,2,4,5,2,5,7,3,6,3,3,54,66,98,3,6,3,5,4,56])
arr
```

```python
q1=np.percentile(arr,25)
q3=np.percentile(arr,75)
iqr=q3-q1
lb=q1-(1.5*iqr)
ub=q3+(1.5*iqr)

print(q1,q3,iqr,lb,ub)
```

```python
li=[]
for i in arr:
    if i >= -1.875 and i<=11.125:
        li.append(i)
arr2=np.array(li)
arr2
```

```python
sns.boxplot(arr)
```

```python
sns.boxplot(x=arr2)
```
