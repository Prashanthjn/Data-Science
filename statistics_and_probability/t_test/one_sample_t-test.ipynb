```python
import scipy.stats as stats
import numpy as np

pop_mean=45
s_mean=75
s_std_dev=25
n=25

t_stat=(s_mean-pop_mean)/(s_std_dev/np.sqrt(n))

dof=n-1
alpha=0.05
critical_value=stats.t.ppf(1-alpha,dof)

p_val=1-stats.t.cdf(t_stat,dof)

print(f"t-statistic:{t_stat}")
print(f"critical value:{critical_value}")
print(f"p-value:{p_val}")

print("with t value")
if t_stat> critical_value:
    print("there is a significant difference in mean(reject H0)")
else:
    print("there is no significant difference(accept H0)")

print("with p value")
if p_val>alpha:
    print("no significant diffrence in means(accept H0)")
else:
    print("there is a significant difference in means(reject H0)")
```
