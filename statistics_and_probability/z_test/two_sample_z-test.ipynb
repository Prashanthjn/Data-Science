```python
null_hypothesis="there is no significant diffrence in means of these two groups"
import numpy as np 
import scipy.stats as stats

n1=50
x1=75
sigma_1=10

n2=60
x2=80
sigma_2=12

d=0   # we assume that there is no diffrence in means
alpha=0.05

print(f"null hypothesis:{null_hypothesis}")
z_score =((x1-x2)-d) / np.sqrt((sigma_1**2/n1) + (sigma_2**2/n2))

print(f"z-score:{z_score}")

critical=stats.norm.ppf(1-alpha/2)
print(f"critical value :{critical}")

if np.abs(z_score) > critical:
    print("reject null hypothesis")
else:
    print("accept null hypothesis")
```
