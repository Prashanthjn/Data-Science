```python
import numpy as np 
from statsmodels.stats.weightstats import ztest

data=np.random.normal(loc=11.8, scale=0.5,size=100)
pop_mean=12
samp_mean=11.5

z_statistic,p_value=ztest(data,value=pop_mean)

print(f"z-statistic:{z_statistic:.4f}")
print(f"p-value:{p_value:}")

alpha=0.05
if p_value<alpha:
    print("Reject the null hypothesis (H0) : the means are diffrent")
else:
    print("failed to reject the null hypothesis (H0): the means are same")
```
